```python
tabla = int(input("Que tabla de multiplicar quieres?"))
limiteinicial = int(input("Limite inicial"))
limitefinal = int(input("Limite final"))
while limitefinal <= limiteinicial:
  print("Error: El limite final debe ser mayor que el inicial")
  limitefinal = int(input("Limite final"))
incrementos = int(input("Dame el numero de incrementos"))
for i in range(limiteinicial, limitefinal, incrementos):
  print(tabla, "*", i, "=", tabla*i)
```
